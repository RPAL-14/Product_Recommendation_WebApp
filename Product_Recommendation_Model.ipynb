```python
import pandas as pd
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

# Load data
df = pd.read_csv("/Users/rudraanpal/Desktop/IKARUS3D_ASSIGNMENT/intern_data_ikarus.csv")

# Fill missing text fields with ""
for col in ['title', 'brand', 'description', 'categories']:
    df[col] = df[col].fillna("")

# Combine text fields for embedding
df['combined_text'] = df['title'] + " " + df['brand'] + " " + df['categories'] + " " + df['description']

# Load SentenceTransformer model (use a small one for speed, e.g., "all-MiniLM-L6-v2")
model = SentenceTransformer('all-MiniLM-L6-v2')

# Compute embeddings for all products
embeddings = model.encode(df['combined_text'].tolist(), show_progress_bar=True)

# Compute cosine similarity matrix
similarity_matrix = cosine_similarity(embeddings)

def recommend_products(product_id, top_n=5):
    # Get similarity scores for the product_id (index in dataframe)
    similarity_scores = list(enumerate(similarity_matrix[product_id]))
    # Sort by similarity descending, skip first which is the product itself
    similarity_scores = sorted(similarity_scores, key=lambda x: x[1], reverse=True)[1:top_n+1]
    # Get recommended product indices
    recommended_ids = [i[0] for i in similarity_scores]
    return df.iloc[recommended_ids][['title', 'brand', 'categories', 'price', 'description']]

# Example: recommend for product at index 0
print(recommend_products(0))
```
